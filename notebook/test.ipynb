```python
import os
import re
import pickle
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import json
import glob
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
pd.set_option('display.max_colwidth', None)
```

```python
json_dir = '../results/2021/'
flattened_data = []
json_files = glob.glob(os.path.join(json_dir, '*.json'))
for json_file in json_files: 
    with open(json_file, 'r') as f:
        data = json.load(f)
        for author_data in data:
            author = author_data['author']
            for window in author_data['windows']:
                record = {
                    'author': author,
                    'start_index': window['start_index'],
                    'start_time': window['start_time'],
                    'end_index': window['end_index'],
                    'end_time': window['end_time'],
                    'timezones': window['timezones']
                }
                flattened_data.append(record)

df = pd.DataFrame(flattened_data)
```

```python
with open('windows_2021.pkl', 'wb') as f:
    pickle.dump(df, f)
```

```python
# Convert datetime strings to pandas datetime objects when loading the data
df['start_time'] = pd.to_datetime(df['start_time'])
df['end_time'] = pd.to_datetime(df['end_time'])

# Calculate duration in hours more efficiently
df['duration'] = (df['end_time'] - df['start_time']).dt.total_seconds() / 3600

# Updated author_stats calculation
author_stats = pd.DataFrame({
    'total_windows': df.groupby('author').size(),
    'unique_patterns': df.groupby('author')['timezones'].apply(lambda x: len(set(tuple(tz) for tz in x))),
    'avg_changes_per_window': df.groupby('author')['timezones'].apply(lambda x: np.mean([len(tz) for tz in x])),
    'window_durations': df.groupby('author')['duration'].mean()
})
author_stats.head(2)
```
